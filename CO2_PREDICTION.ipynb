```python
import math



def tinh(N):
  #Dien tich - cm2
  A = 365

  #The tich CO2 final - cm3 or mL
  SQC = 2100
  #Nhiet do - oC
  T = 25.578
  #The tich headspace - cm3 or mL
  VH = 21.2
  #The tich chat long - cm3 or mL
  VL = 500
  #Crystallinity - %
  CX = 60
  #Wall thickness - cm
  XW = 0.0508

  KD = 1.467*(10**-2)*(math.exp(936/(273+T)))
  CH = 40.331 - 0.1107*(273+T)
  B  = 361.6/(273+T) - 0.8558

  HENRY_AP =  (1/((math.e**(14-0.013341*(273+T)-558.98/(273+T)-422577/((273+T)**2)))))/0.986923
  DH = 8.376*(10**-3)*(math.exp(-5414/(273+T)))
  DD = 9.36*(10**-3)*(math.exp(-4596/(273+T)))
  HL = 4.109*(10**9)*(math.exp(-2425/(273+T)))


  R_value = 62358.974359
  G = DH/DD
  K = CH*B/KD
  target = 0.85*SQC

  time, node = 11001110, N+2
  C = [[0 for x in range(time)] for y in range(node)]

  for j in range(0, len(C[0])):
    C[N+1][j] =  0

  for j in range(0, len(C)):
    C[j][0] =  0
  #DentaX nè - cm
  DELX = XW/N

  j = 1
  tau_congdon = 0
  tau =0
  while SQC > target:
    tau_congdon = tau_congdon + tau
    for i in range(1,N):
      P = (SQC*0.001/22.4)/((VH/1000)/(8.314*(T+273))+ HENRY_AP*VL/1000)
      C[0][j] = KD*P+(CH*B*P)/(1+B*P)
      C[i][j] = (C[i-1][j-1]+C[i+1][j-1])/2
      Y = G*K/((1+B*P)**2)
      U = K/((1+B*P)**2)
      DEFF = DD*(1+Y)/(1+U)
      tau = (DELX**2)/(2*DEFF)
      QCP = DEFF*A*(C[0][j]-C[1][j])*tau/DELX
    SQC = SQC - QCP
    j+=1

  print(tau_congdon*1.65344e-6)
  return tau_congdon*1.65344e-6

k1 = tinh(6)
```
